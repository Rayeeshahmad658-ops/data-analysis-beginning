```python
# Student Marks Analysis
```

```python
import pandas as pd

data = {
  "Name": ["A", "B", "C", "D"],
  "Marks": [80,73,64,90]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.describe())
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```
