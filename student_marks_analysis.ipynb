```python
import pandas as pd

data = {
  "Name": ["A", "B", "C", "D", "e"],
  "Marks": [80,67,73,64,90]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.describe())
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import pandas as pd

# Calculate the average marks
average_marks = df["Marks"].mean()
print(f"The average mark is: {average_marks}")
```

```python
# 1. Plot the bar chart for student marks
plt.bar(df["Name"], df["Marks"], color="skyblue", label="Student Marks")

# 2. Add the horizontal line at the average mark (74.8)
plt.axhline(y=74.8, color="red", linestyle="--", linewidth=2, label="Average (74.8)")

# 3. Add labels and a legend to make it look clean
plt.title("Student Marks Distribution")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.legend()

# 4. Display the complete plot
plt.show()
```

```python
# Find the student with the highest and lowest marks
highest_student = df.loc[df["Marks"].idxmax()]
lowest_student = df.loc[df["Marks"].idxmin()]

print(f"Top Student: {highest_student['Name']} with {highest_student['Marks']} marks")
print(f"Needs Improvement: {lowest_student['Name']} with {lowest_student['Marks']} marks")
```

```python
# Filter for students below average
below_average = df[df["Marks"] < 74.8]
print("Students scoring below average:")
print(below_average[["Name", "Marks"]])
```

```python
# Get a summary table of the data
df.describe()
```

```python
# 1. Define the grade boundaries and corresponding labels
bins = [0, 65, 75, 85, 100]
labels = ['F', 'C', 'B', 'A']

# 2. Create a new 'Grade' column
df['Grade'] = pd.cut(df['Marks'], bins=bins, labels=labels)

# 3. View the final, updated DataFrame
print("Final Student Grading Sheets:")
print(df)

# 4. Save your hard work to a new CSV file
df.to_csv("student_marks_final_summary.csv", index=False)
print("\n Results successfully saved to 'student_marks_final_summary.csv'!")
```
